# Flow Results Visualization

Regenerates the flow comparison results table and renders it as a PNG image.

| Step | Script | Output |
|------|--------|--------|
| 1 — compile results | `export_results_table.py` | `results/flow_comparison_table.csv` |
| 2 — render table | this notebook | `graphics/flow_comparison_table.png` |

Import libraries for file operations, tabular data loading, subprocess execution, and figure generation.

```python
import os
import subprocess
import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path("../").resolve()
```

Define the path to the results CSV and the destination path for the output PNG.

```python
DATA_PATH   = "../results/flow_comparison_table.csv"
OUTPUT_PATH = "../graphics/flow_comparison_table.png"
```

Run `export_results_table.py` to compile all experiment results (including
`24hr_flow_all_inputs_noSTG`) into `results/flow_comparison_table.csv`.
This ensures the CSV is always up-to-date before rendering.

```python
result = subprocess.run(
    [sys.executable, "export_results_table.py"],
    cwd=ROOT,
    capture_output=False,
)
if result.returncode != 0:
    raise RuntimeError(f"export_results_table.py exited with code {result.returncode}")
```

Load the results CSV and preview it as a DataFrame to confirm the data before plotting.

```python
df = pd.read_csv(DATA_PATH)
df
```

Render the DataFrame as a matplotlib table on a blank axis, apply font and scale styling, then save the figure to the `graphics/` folder (creating it if needed) and display it inline.

```python
fig, ax = plt.subplots(figsize=(11, max(4, len(df) * 0.28 + 1)))
ax.axis("off")

table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    cellLoc="center",
    loc="center",
)
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.2, 1.2)

os.makedirs(os.path.dirname(OUTPUT_PATH), exist_ok=True)
plt.savefig(OUTPUT_PATH, bbox_inches="tight", dpi=300)
print(f"Saved \u2192 {OUTPUT_PATH}")
plt.show()
```
